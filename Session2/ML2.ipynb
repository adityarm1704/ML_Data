```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score,root_mean_squared_error
```

```python
df=pd.read_csv("Advertising.csv")
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
#checking  linear relationship
plt.figure()
plt.scatter(df['TV'],df['Sales'])
plt.show()
```

```python
plt.figure()
plt.scatter(df['Newspaper'],df['Sales'])
plt.show()
```

```python
df.corr()
```

```python
sns.heatmap(df.corr(),annot=True)
```

```python
df.drop("Unnamed: 0",axis=1,inplace=True)
```

```python
df.head()
```

```python
# seperate input and ouput feature
X=df[['TV']] # independent value
Y=df[["Sales"]] # dependent value
```

```python
# perform splitting of training and testing data
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
#create an object of linear regression
lr=LinearRegression() # this object is going to call OLS(ordinary least square function) which is used to give best theta1 and theta0 value.
```

```python
lr.fit(X_train,Y_train)
```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
# calculate predicted value on training data
Y_predict=lr.predict(X_train)
```

```python
Y_predict
```

```python
plt.figure()
plt.scatter(X_train,Y_train)
plt.plot(X_train,Y_predict)
plt.show()
```

```python
#now perform prediction on testing data
y_pred=lr.predict(X_test)
```

```python
# Model evaluation
# mse=mean_squared_error
mse=mean_squared_error(y_pred,Y_test)
```

```python
rmse=root_mean_squared_error(y_pred,Y_test)
```

```python
mae=mean_absolute_error(y_pred,Y_test)
```

```python
print(mse)
print(rmse)
print(mae)
```
